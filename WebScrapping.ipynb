```python
#start by importing our packages
from bs4 import BeautifulSoup
import requests
```

```python
#the website I will be working on
url = 'https://books.toscrape.com/catalogue/category/books_1/index.html'
html_text = requests.get(url).text
print(html_text)
```

```python
#this code allows to just pick one book from the list on the first page
# Extract the HTML content from the Response object using .text
soup = BeautifulSoup(html_text, "html.parser")  # Changed: Added .text to get HTML content from Response object

book = soup.find('article', class_='product_pod')

title = book.find('h3').find('a')['title']
print(title)
```

```python
#code shows all the listed book titles and their prices
html_text = requests.get(url)
html_text.encoding = 'utf-8' 
soup = BeautifulSoup(html_text.text, "html.parser")

books = soup.find_all('article', class_='product_pod')

for book in books:
    title = book.find('h3').find('a')['title']
    price = book.find('p', class_='price_color').text
    print(f'Title: {title}')
    print(f'Price: {price}')
    print('-' * 50)

    
```

```python
#listing different book genres
genre_list = soup.find('ul', class_ = 'nav nav-list')
genre = genre_list.find_all('a')
for genre in genre:
    print(genre.text.strip())
```

```python
#finding books under a specific genre
genre_list = soup.find('ul', class_ = 'nav nav-list')
sub_genres = genre_list.find('ul').find_all('a')

for genre in sub_genres:
    if genre.text.strip() == 'Travel':
        travel_link = genre['href']
        
        print(travel_link)
```

```python
#all books under travel genre and their prices
base_url = 'https://books.toscrape.com/catalogue/category/books/'
travel_url = base_url + travel_link

travel_html = requests.get(travel_url)
travel_html.encoding = 'utf-8'  # force correct encoding
travel_soup = BeautifulSoup(travel_html.text, "html.parser")

books = travel_soup.find_all('article', class_='product_pod')

for book in books:
    title = book.find('h3').find('a')['title']
    price = book.find('p', class_='price_color').text.replace('\xa3', '£')
    print(f'Ttile: {title}')
    print(f'Price: {price}')
    print('-' * 50)
    
```
